```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
df=pd.read_csv(r"C:\Users\\Downloads\2. Finance_Analytics_Dataset.csv")
df
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x="Category")
plt.title("Number of Transactions in Each Category")
plt.xlabel("Category")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x="PaymentMode")
plt.title("Number of Transactions by Payment Mode")
plt.xlabel("Payment Mode")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="TransactionType")
plt.title("Number of Transactions by Transaction Type")
plt.xlabel("Transaction Type")
plt.ylabel("Number of Transactions")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(
    data=df,
    x="TransactionType",
    y="Amount",
    estimator=sum
)
plt.title("Debit vs Credit Amount Comparison")
plt.xlabel("Transaction Type")
plt.ylabel("Total Amount")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 6))
sns.barplot(
    data=df,
    x="CustomerSegment",
    y="Amount",
    estimator=sum
)
plt.title("Spending Pattern by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Transaction Amount")
plt.tight_layout()
plt.show()
```

```python
payment_counts = df["PaymentMode"].value_counts()

plt.figure(figsize=(8, 8))
plt.pie(
    payment_counts,
    labels=payment_counts.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Payment Mode Usage Distribution")
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.boxplot(
    data=df,
    x="Category",
    y="Amount"
)
plt.title("Boxplot of Transaction Amount by Category")
plt.xlabel("Category")
plt.ylabel("Transaction Amount")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
