```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
df=pd.read_csv(r"C:\Users\Reddy Sai Abhiram\Downloads\house.csv")
df
```

```python
X = df['House Size'].to_frame()
Y = df['House Price']
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size=0.2, random_state=42)
```

```python
model = LinearRegression()
model.fit(X_train, Y_train)
```

```python
Y_pred = model.predict(X_test) #testing 
Y_pred
```

```python
r2_score = r2_score(Y_test, Y_pred)
r2_score
#print evoluation metrics
print (f"R-squared Score: ",r2_score)
```

```python
#plot
plt.scatter(X, Y, color="blue", label="Data point")  #All data points
plt.plot(X, model.predict(X), color="red", label="Regression Line") # Regression line
plt.xlabel("House Size (sqft)")
plt.ylabel("House Price ($)")
plt.title("House Price vs. Size")
plt.legend()
plt.show()
```

```python
# User input for House Size
user_House_Size = int(input("Enter House Size to predict salary: "))

# predicted House Price for House Size
user_df = pd.DataFrame([[user_House_Size]], columns=['House Size'])

predicted_salary = model.predict(user_df)
print(f"The predicted House Price for House Size (user_House_Size) is:",predicted_salary)
```
